```python
#やまもとそうま

import cv2
import numpy as np
import glob
import os

print("--- 山本そうま100x100 メッシュ RGB分布解析ツール ---")
print("ファイル名      | 良品色(暗い色)の割合 | Rばらつき | Gばらつき | Bばらつき")
print("-" * 75)

# 1. ★ここがポイント！画像が入っているフォルダの正しいパスを指定する
# notebooksフォルダから見て、1つ上(../)のdata/processed/...を指定します
target_folder = "../data/processed/Yamamoto/20260717/"

# 2. 指定したフォルダの中にある画像を取得するように変更
all_images = list(set(
    glob.glob(target_folder + "*.jpg") + 
    glob.glob(target_folder + "*.JPG") + 
    glob.glob(target_folder + "*.jpeg") + 
    glob.glob(target_folder + "*.png")
))

# 解析対象にする画像を指定（今回は "FAMACsPb" から始まる画像）
image_files = [f for f in all_images if os.path.basename(f).startswith("FAMACsPb")]
image_files.sort()

if len(image_files) == 0:
    print(f"⚠️ エラー: '{target_folder}' の中に 'FAMACsPb' から始まる画像が見つかりません。")
    print("-" * 75)

for img_path in image_files:
    img = cv2.imread(img_path)
    if img is None:
        continue

    # 1. 100x100のメッシュ化（縮小して平均化）
    mesh_size = 100
    mesh_img = cv2.resize(img, (mesh_size, mesh_size), interpolation=cv2.INTER_AREA)

    # 2. BGR から RGB に変換
    rgb_img = cv2.cvtColor(mesh_img, cv2.COLOR_BGR2RGB)
    
    # R, G, B それぞれの色ごとにデータを分割
    r, g, b = cv2.split(rgb_img)

    # 3. 各色の「ばらつき（標準偏差）」を計算
    r_std = np.std(r)
    g_std = np.std(g)
    b_std = np.std(b)

    # 4. 「1万マスに対する特定の色の割合」を計算
    good_pixels = np.sum((r < 80) & (g < 80) & (b < 80))
    
    # 全体に対する良品マスの割合（%）を算出
    total_pixels = mesh_size * mesh_size
    good_ratio = (good_pixels / total_pixels) * 100

    # 結果を出力（os.path.basenameでファイル名だけを綺麗に表示）
    filename = os.path.basename(img_path)
    print(f"{filename:30} | {good_ratio:6.1f}%             | R:{r_std:5.1f} | G:{g_std:5.1f} | B:{b_std:5.1f}")

print("-" * 75)
```
